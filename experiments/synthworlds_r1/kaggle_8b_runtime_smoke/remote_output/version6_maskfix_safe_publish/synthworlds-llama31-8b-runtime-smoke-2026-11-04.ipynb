It reads no research corpus, creates no research result, and does not start SFT/unlearning.

First, turn Internet on in Notebook Settings and enable the existing `HF_TOKEN` secret under Add-ons → Secrets. Set `RUN_ACCESS_PREFLIGHT = True` and run the cell. The token is never printed or saved. If the receipt says `PASS_HF_ACCESS_PREFLIGHT`, select GPU T4 x2, set `RUN_ACCESS_PREFLIGHT = False` and `RUN_SMOKE = True`, then run the cell again.
